# 01 — Exploration des données

Exploration des snapshots JSON collectés par le collecteur PRIM.

```python
import json
import pandas as pd
from pathlib import Path

# Charger un snapshot exemple
sample = Path('../data/samples/sample_snapshot.json')
with open(sample) as f:
    data = json.load(f)

print(f"fetched_at: {data['fetched_at']}")
print(f"Nombre d'appels: {len(data['metro_calls'])}")
```

```python
df = pd.DataFrame(data['metro_calls'])
print(df.dtypes)
df.head()
```
